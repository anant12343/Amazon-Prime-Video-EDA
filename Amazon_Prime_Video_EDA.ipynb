### Executive Summary

This project is a business-oriented exploratory data analysis (EDA) of the Amazon Prime Video catalogue available in the United States. It uses two files: `titles.csv` (9,871 rows, 15 columns: one row per title) and `credits.csv` (124,235 rows, 5 columns: one row per actor/director credit). After removing 3 exact duplicate titles and the 56 duplicate credit rows linked to them, the analysis works on **9,868 unique titles** and **124,179 credits**. The work was done in Python with Pandas, NumPy, Matplotlib and Seaborn.

**Approach.** Following the project architecture, assumptions are listed first, tested with 28 visualisations in UBM order (univariate, bivariate, multivariate) and simple statistics, then converted into business insights. Two columns (`genres`, `production_countries`) store lists as text; they were parsed and "exploded" so that a title with three genres counts once in each genre. Missing ratings were **not** imputed, because inventing a rating would distort every average; each statistic instead reports how many titles it is based on.

**What the data shows.**
- **Movie-heavy library:** 86.2% of titles are movies and 13.8% are TV shows.
- **Recent-release skew:** the median release year is 2014 and 60.4% of titles were released in 2010 or later, yet 14.3% were released before 1970 - a large classic/archive long tail.
- **Genre concentration:** drama is tagged on 48.3% of titles, comedy on 30.3% and thriller on 21.5%. Two in three titles carry more than one genre.
- **Quality varies sharply by genre:** documentaries (mean IMDb 6.99) and history (6.81) score highest; horror (4.77) and sci-fi (5.37) score lowest even though horror is in the top-10 genres by volume.
- **TV shows rate higher than movies:** mean IMDb 7.12 vs 5.80 (median 7.3 vs 5.9).
- **US dominates volume, not ratings:** the US appears on 54.0% of titles, followed by India (10.9%) and the UK (9.4%). Among countries with 100+ titles, Japan (6.66), China (6.63) and the UK (6.42) have the highest average IMDb scores, while US (5.80) and Canada (5.77) are at the bottom. Non-US titles average 0.48 points higher than US-only titles.
- **Quality is not the same as reach:** the correlation between IMDb score and IMDb votes is weak (Spearman 0.14). More than half (53.5%) of titles rated 8.0+ have fewer than 1,000 votes - a pool of 606 "hidden gems" (score >= 7.5, < 1,000 votes) exists.
- **Cast data reflects the archive:** the most-credited actors (e.g., George 'Gabby' Hayes, Roy Rogers, Gene Autry) and directors (Joseph Kane, Sam Newfield) are mostly from 1930s-50s US westerns, and their titles collect very few votes. There are 75,200 unique actors and 6,304 unique directors.

**Business interpretation (data-driven implications, not statements of Amazon's actual strategy).** The catalogue has breadth (116 production countries, 19 genres) but its largest segments - US movies, thrillers, horror and action - are also its lowest-rated. Higher-rated segments (documentaries, TV series, Indian, Japanese and UK content) are smaller. Possible considerations are: rebalancing investment toward higher-rated genres and formats, improving discovery for highly rated but low-visibility titles, treating the classic archive as a niche asset rather than a growth driver, and using country-level quality signals for regional sourcing.

**Limitations.** `release_year` is the original release year, **not** the date a title was added to Prime Video, so year trends describe the catalogue's age profile, not Amazon's acquisition history. The data is a single snapshot (2022 is partial) with no viewership, revenue or "Amazon Original" information. IMDb/TMDB metrics reflect those sites' users, and all relationships reported are correlations, not causation.

### Helper Functions

Re-usable functions keep the notebook modular: loading with error handling, parsing list-like text columns, reshaping lists into rows, consistent chart formatting, and two small NumPy-only statistics helpers.

````python
# ---------- Data helpers ----------
def load_csv(path):
    """Load a CSV file and report its size. Raises a clear message if the file is missing."""
    try:
        df = pd.read_csv(path)
        print(f'Loaded {path}: {df.shape[0]:,} rows x {df.shape[1]} columns')
        return df
    except FileNotFoundError:
        raise FileNotFoundError(
            f"'{path}' was not found. In Colab, upload the file using the Files panel "
            "(left sidebar) or set DATA_DIR to the folder that contains it.")


def parse_list_column(series):
    """Turn strings such as "['drama', 'comedy']" into Python lists.
    Anything that cannot be parsed (missing / malformed) becomes an empty list."""
    def _safe_parse(value):
        try:
            parsed = ast.literal_eval(value)
            return list(parsed) if isinstance(parsed, (list, tuple)) else []
        except (ValueError, SyntaxError, TypeError):
            return []
    return series.apply(_safe_parse)


def explode_list_column(df, list_col, new_col):
    """Return a 'long' table with one row per (title, list item).
    Example: a title tagged ['drama', 'crime'] becomes two rows."""
    long_df = df.explode(list_col).rename(columns={list_col: new_col})
    return long_df[long_df[new_col].notna()].reset_index(drop=True)


# ---------- Chart helpers ----------
def finish_plot(ax, title, xlabel='', ylabel='', keep_legend=False):
    """Apply the same title / label / spine formatting to every chart."""
    ax.set_title(title, loc='left', pad=12)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    sns.despine(ax=ax)
    if not keep_legend and ax.get_legend() is not None:
        ax.get_legend().remove()


def label_bars(ax, fmt='{:,.0f}', fontsize=9):
    """Write the value at the end of every bar (works for vertical and horizontal bars)."""
    for container in ax.containers:
        labels = [fmt.format(v) for v in container.datavalues]
        ax.bar_label(container, labels=labels, padding=3, fontsize=fontsize, color=TEXT_MUTED)


# ---------- Statistics helpers (NumPy only) ----------
def bootstrap_mean_diff(a, b, n_boot=2000, seed=42):
    """Difference in means (a - b) with a 95% bootstrap confidence interval.
    We resample each group with replacement many times and look at the spread of the differences."""
    rng = np.random.default_rng(seed)
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    diffs = np.array([rng.choice(a, a.size).mean() - rng.choice(b, b.size).mean() for _ in range(n_boot)])
    return a.mean() - b.mean(), np.percentile(diffs, [2.5, 97.5])


def cohens_d(a, b):
    """Effect size: difference in means divided by the pooled standard deviation.
    Rough guide: 0.2 = small, 0.5 = medium, 0.8 = large."""
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    pooled_sd = np.sqrt(((a.size - 1) * a.var(ddof=1) + (b.size - 1) * b.var(ddof=1)) / (a.size + b.size - 2))
    return (a.mean() - b.mean()) / pooled_sd

print('Helper functions ready.')
````

## 3. ***Data Wrangling***

### Cleaning decisions (explained before implementing)

| # | Issue found | Decision | Why |
|---|---|---|---|
| 1 | 3 exact duplicate title rows and 56 duplicate credit rows | Drop exact duplicates | They are copies; keeping them double-counts titles, genres and credits |
| 2 | `genres` and `production_countries` are lists stored as text | Parse into real lists with `ast.literal_eval` | Needed to count each genre / country correctly |
| 3 | Titles with several genres or countries | Create "long" tables (one row per title-genre and title-country). A title counts **once in each** of its genres/countries | Keeps every tag; percentages are "% of titles that carry this tag", so they can add up to more than 100% |
| 4 | Country code `XX` (120 titles) and empty country lists | Remove `XX` from lists; titles with no known country are labelled **Unknown** and excluded from country-level comparisons | `XX` is a placeholder, not a country |
| 5 | Missing IMDb / TMDB scores, votes, popularity | **Not imputed**; each analysis uses titles where the value exists and reports n | Filling ratings with a mean/median would invent quality information and bias averages |
| 6 | `seasons` missing for movies | Leave as NaN (structural) | Seasons do not apply to movies |
| 7 | `age_certification` ~66% missing | Fill with "Not available" for counting only | Too much missing to impute; dropping would remove most of the data |
| 8 | `description` / `character` missing | Fill with empty / "Unknown" text | Not used in numeric analysis |
| 9 | Heavy skew in `imdb_votes` and `tmdb_popularity`, extreme runtimes | **Keep** outliers; use medians, log scales and log-transformed columns | Blockbusters and long titles are real data, not errors |
| 10 | Ratings based on very few votes are unstable | Use minimum-vote thresholds (1,000 or 10,000 votes) when ranking titles or comparing genres | A 9.7 from 30 votes is not comparable with an 8.8 from 400,000 votes |

**Derived columns:** `genre_list`, `country_list`, `n_genres`, `n_countries`, `is_coproduction`, `origin` (US only / US co-production / Non-US / Unknown), `decade`, `log_votes`, `log_popularity`.

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Step 1 - remove exact duplicates (work on copies so raw data stays untouched)
titles = titles_raw.drop_duplicates().copy()
credits = credits_raw.drop_duplicates().copy()
assert titles['id'].is_unique, 'Title IDs should be unique after removing duplicates'
print(f'titles : {len(titles_raw):,} -> {len(titles):,} rows')
print(f'credits: {len(credits_raw):,} -> {len(credits):,} rows')

# Step 2 - parse list-like text columns into real lists
titles['genre_list'] = parse_list_column(titles['genres'])
titles['country_list'] = parse_list_column(titles['production_countries'])

# Step 4 - drop the placeholder country code 'XX'
titles['country_list'] = titles['country_list'].apply(lambda codes: [c for c in codes if c != 'XX'])

# Helper counts
titles['n_genres'] = titles['genre_list'].str.len()
titles['n_countries'] = titles['country_list'].str.len()
titles['is_coproduction'] = titles['n_countries'] >= 2

# Origin group used for the US vs other-countries comparison
is_us_only = titles['country_list'].apply(lambda codes: codes == ['US'])
has_us = titles['country_list'].apply(lambda codes: 'US' in codes)
titles['origin'] = np.select(
    [titles['n_countries'] == 0, is_us_only, has_us],
    ['Unknown', 'US only', 'US co-production'],
    default='Non-US')

# Step 7 and 8 - fill text columns used only for labelling
titles['age_certification'] = titles['age_certification'].fillna('Not available')
titles['description'] = titles['description'].fillna('')
credits['character'] = credits['character'].fillna('Unknown')

# Time and log-scale helper columns (log10 of 0 is undefined, so popularity uses +1)
titles['decade'] = (titles['release_year'] // 10) * 10
titles['log_votes'] = np.log10(titles['imdb_votes'])
titles['log_popularity'] = np.log10(titles['tmdb_popularity'] + 1)

# Step 3 - long tables: one row per title-genre and per title-country
genres_long = explode_list_column(titles, 'genre_list', 'genre')
countries_long = explode_list_column(titles, 'country_list', 'country')

# Credits joined to title attributes, and split by role
credits_full = credits.merge(
    titles[['id', 'title', 'type', 'release_year', 'imdb_score', 'imdb_votes', 'country_list']],
    on='id', how='left')
actors = credits_full[credits_full['role'] == 'ACTOR']
directors = credits_full[credits_full['role'] == 'DIRECTOR']

print('\nTitles with no genre          :', (titles['n_genres'] == 0).sum())
print('Titles with no known country  :', (titles['n_countries'] == 0).sum())
print('Titles with 2+ genres         :', f"{(titles['n_genres'] >= 2).mean():.1%}")
print('Co-productions (2+ countries) :', titles['is_coproduction'].sum())
print('genres_long rows              :', f'{len(genres_long):,}', '| unique genres   :', genres_long['genre'].nunique())
print('countries_long rows           :', f'{len(countries_long):,}', '| unique countries:', countries_long['country'].nunique())
print('\nOrigin groups:'); print(titles['origin'].value_counts())
````

````python
# Chart - 4 visualization code
# Question: how concentrated is audience attention (votes / popularity) across titles?
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Log-spaced bins are needed because both variables span several orders of magnitude
votes = titles['imdb_votes'].dropna()
bins_v = np.logspace(np.log10(votes.min()), np.log10(votes.max()), 40)
axes[0].hist(votes, bins=bins_v, color=PRIMARY, alpha=0.85, edgecolor='white')
axes[0].set_xscale('log')
axes[0].axvline(votes.median(), color=SECONDARY, linestyle='--', linewidth=2, label=f'median = {votes.median():,.0f}')
axes[0].axvline(votes.mean(), color='#0b0b0b', linestyle=':', linewidth=2, label=f'mean = {votes.mean():,.0f}')
axes[0].legend(frameon=False)
finish_plot(axes[0], 'IMDb votes per title', 'IMDb votes (log scale)', 'Number of titles', keep_legend=True)

pop = titles['tmdb_popularity'].dropna()
pop = pop[pop > 0]   # log scale cannot show zero
low = pop.quantile(0.001)   # ignore a handful of near-zero values that would stretch the axis
bins_p = np.logspace(np.log10(low), np.log10(pop.max()), 40)
axes[1].hist(pop, bins=bins_p, color=PRIMARY, alpha=0.85, edgecolor='white')
axes[1].set_xscale('log')
axes[1].axvline(pop.median(), color=SECONDARY, linestyle='--', linewidth=2, label=f'median = {pop.median():.2f}')
axes[1].axvline(pop.mean(), color='#0b0b0b', linestyle=':', linewidth=2, label=f'mean = {pop.mean():.2f}')
axes[1].legend(frameon=False)
finish_plot(axes[1], 'TMDB popularity per title', 'TMDB popularity (log scale)', 'Number of titles', keep_legend=True)
plt.tight_layout(); plt.show()

# Concentration of attention: share of all votes captured by the top 1% / 10% of titles
v_sorted = votes.sort_values(ascending=False)
for share in [0.01, 0.10]:
    k = int(len(v_sorted) * share)
    print(f'Top {share:.0%} of titles hold {v_sorted.iloc[:k].sum() / v_sorted.sum():.1%} of all IMDb votes')
print(f'Titles with < 1,000 votes: {(votes < 1000).mean():.1%}')
print('Skew - votes:', round(votes.skew(), 1), '| popularity:', round(titles["tmdb_popularity"].skew(), 1))
````

#### Chart - 14 - IMDb score by content type (violin plot + statistical test)

````python
# Chart - 14 visualization code
# Question: is the gap between TV shows and movies large and reliable?
rated = titles.dropna(subset=['imdb_score'])
fig, ax = plt.subplots(figsize=(8, 5))
sns.violinplot(data=rated, x='type', y='imdb_score', palette=TYPE_COLORS, inner='quartile',
               cut=0, linewidth=1, ax=ax)
finish_plot(ax, 'TV shows are rated higher than movies across the whole distribution', 'Content type', 'IMDb score')
plt.tight_layout(); plt.show()

show_scores = rated.loc[rated['type'] == 'SHOW', 'imdb_score']
movie_scores = rated.loc[rated['type'] == 'MOVIE', 'imdb_score']
diff, ci = bootstrap_mean_diff(show_scores, movie_scores)
print(f'Mean SHOW = {show_scores.mean():.2f} (n={len(show_scores):,}) | Mean MOVIE = {movie_scores.mean():.2f} (n={len(movie_scores):,})')
print(f'Difference = {diff:.2f} points, 95% bootstrap CI [{ci[0]:.2f}, {ci[1]:.2f}]')
print(f"Cohen's d = {cohens_d(show_scores, movie_scores):.2f}")

# Check that the gap is not only because shows have more votes: compare titles with >= 1,000 votes
reliable = rated[rated['imdb_votes'] >= 1000]
print('\nTitles with >= 1,000 votes:')
print(reliable.groupby('type')['imdb_score'].agg(['count', 'mean', 'median']).round(2))
````

### Limitations of the analysis
1. **Release year is not date added.** The data cannot show when Amazon added titles, so "trends over time" describe the age profile of the current catalogue.
2. **Single snapshot.** Popularity and votes are point-in-time values; 2022 is only partly covered.
3. **No business outcome data.** There is no viewing time, subscriber, revenue, cost or "Original" flag, so the analysis measures external ratings and attention, not actual Prime Video performance.
4. **Third-party ratings.** IMDb and TMDB ratings reflect those platforms' users and contain self-selection bias; small-vote ratings are unreliable (handled with vote thresholds).
5. **Missing data.** About 10% of titles have no IMDb score, 21% no TMDB score and 66% no age certification. These were not imputed, so results describe titles with available data.
6. **Multi-label counting.** Genres and countries are counted once per tag, so percentages can exceed 100% and genre/country groups overlap.
7. **Correlation is not causation.** All relationships reported (e.g., TV vs rating, country vs rating) are associations and may be driven by other factors (age of title, selection of what gets licensed, genre mix).

# **Conclusion**

1. **The catalogue is broad but uneven.** It covers 116 production countries and 19 genres, but it is dominated by movies (86.2%), US production (54.0% of titles) and mainstream genres (drama on 48.3% of titles).
2. **Volume and quality do not line up.** The largest segments - US-only titles, movies, thrillers, action and horror - are below the catalogue's average IMDb score, while smaller segments - TV shows, documentaries, history, and content from India, the UK, Japan and China - rate higher.
3. **TV shows are the strongest format.** They average 7.12 vs 5.80 for movies, and the advantage holds in every genre.
4. **Quality and attention are separate dimensions.** IMDb score is only weakly related to votes and TMDB popularity; 606 highly rated titles have fewer than 1,000 votes, which is a discoverability opportunity.
5. **The library has a large classic archive.** About 1 in 7 titles predates 1970, and the most-credited actors, directors and partnerships come from 1930s-50s westerns, which attract little audience attention today.
6. **Recent Indian content stands out**, appearing prominently among top-rated titles and among the highest-rated prolific actors.

**Most important insight:** in this catalogue, the content that exists in the largest volume is not the content audiences rate highest. Rebalancing toward higher-rated formats, genres and regions, and making existing high-quality titles easier to discover, are the main data-supported opportunities.

**Future improvements:** add viewing/engagement data and an "Original vs licensed" flag; track the catalogue over several snapshots to measure real additions and removals; build a dashboard for content teams; and test predictive models (e.g., predicting rating or popularity from genre, country, cast and runtime).
